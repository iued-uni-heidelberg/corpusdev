```python
import pandas as pd
import xml.etree.ElementTree as ET

def indent(elem, level=0):
    i = "\n" + level * "  "
    if len(elem):
        if not elem.text or not elem.text.strip():
            elem.text = i + "  "
        if not elem.tail or not elem.tail.strip():
            elem.tail = i
        for elem in elem:
            indent(elem, level + 1)
        if not elem.tail or not elem.tail.strip():
            elem.tail = i
    else:
        if level and (not elem.tail or not elem.tail.strip()):
            elem.tail = i

def dataframe_to_xml(df, root_tag, row_tag, filename='output.xml'):
    """
    Converts a pandas DataFrame to an XML file.

    Args:
        df (pd.DataFrame): The input DataFrame.
        root_tag (str): The name for the root XML element.
        row_tag (str): The name for each row XML element.
        filename (str): The name of the output XML file.
    """
    root = ET.Element(root_tag)

    for index, row in df.iterrows():
        row_element = ET.SubElement(root, row_tag)
        for col_name, value in row.items():
            col_element = ET.SubElement(row_element, str(col_name))
            col_element.text = str(value)

    indent(root) # Apply indentation
    tree = ET.ElementTree(root)
    tree.write(filename, encoding='utf-8', xml_declaration=True)
    print(f"DataFrame successfully converted to {filename}")


'''
# Example usage with your 'df' DataFrame
# You might want to select a subset of your DataFrame for XML conversion if it's very large
# For demonstration, let's convert the first 5 rows of selected_columns
example_df = selected_columns.head(5)
dataframe_to_xml(example_df, 'speeches_data', 'speech', 'speeches.xml')

# To view the content of the generated XML file (optional)
with open('speeches.xml', 'r') as f:
    print(f.read())

'''
```
